```python
from transformers import pipeline
from diffusers import StableDiffusionPipeline
from gtts import gTTS
import torch
from IPython.display import display, Audio

# --------------------------------------------------
# Topic
# --------------------------------------------------

topic = "The benefits of renewable energy"


# --------------------------------------------------
# 1. TEXT GENERATION
# --------------------------------------------------

print("Generating text...")

text_generator = pipeline(
    "text2text-generation",
    model="google/flan-t5-base"
)

text_prompt = f"Write a short, engaging paragraph about: {topic}"

generated_text = text_generator(
    text_prompt,
    max_length=80
)[0]["generated_text"]

print("\nGenerated Text:")
print(generated_text)


# --------------------------------------------------
# 2. IMAGE GENERATION
# --------------------------------------------------

print("\nLoading Stable Diffusion model...")

image_prompt = (
    f"An illustration representing {topic}, "
    "digital art, highly detailed"
)

sd_pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
)

sd_pipe = sd_pipe.to("cuda")

print("Generating image...")

image = sd_pipe(
    image_prompt,
    num_inference_steps=25,
    guidance_scale=7.5
).images[0]

image.save("content_image.png")

print("Image saved as content_image.png")

display(image)


# --------------------------------------------------
# 3. AUDIO GENERATION
# --------------------------------------------------

print("\nGenerating audio...")

tts = gTTS(
    text=generated_text,
    lang="en"
)

tts.save("content_audio.mp3")

print("Audio saved as content_audio.mp3")

display(Audio("content_audio.mp3"))
```
